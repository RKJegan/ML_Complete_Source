```python
# importing matplotlib

import matplotlib.pyplot as plt

%matplotlib inline
```

```python
import numpy as np
```

```python
# simple example

x = np.arange(0,10)
y = np.arange(11,21)
```

```python
# Scatter plot

plt.scatter(x,y,c = 'b')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("sample graph")
plt.savefig("test.png")
```

```python
y = x*x
```

```python
# plt plot

#plt.plot(x,y,'g*--')
plt.plot(x,y,'go',linestyle='dashed',linewidth=1,markersize=5)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("sample graph")
```

```python
# creating Sub plots       #--> subplot(row,col,pos)

plt.subplot(2,2,1)            #--> subplot(1row,1col)
plt.plot(x,y,'g--')
plt.subplot(2,2,2)            #--> subplot(1row,2col)
plt.plot(y,x,'r*')
plt.subplot(2,2,3)            #--> subplot(2row,1col)
plt.plot(x,(y**4))
```

```python
# compute x & y for points on a sine curve

x = np.arange(0,4 * np.pi,0.1)
y = np.sin(x)

# plot
plt.plot(x,y)
plt.title("Sine Curve")
```

```python
# Subplot for sine and cosine curve

x = np.arange(0,10 * np.pi, 0.1)
y = np.sin(x)
z = np.cos(x)

#plot sine and cos

plt.subplot(1,2,1)
plt.plot(x,y,'b')
plt.title('Sine Wave')
plt.subplot(1,2,2)
plt.plot(x,z,'y')
plt.title('Cosine Wave')

plt.show()
```

```python
#  Bar Plot

x = [1,2,3,4,5]
y = [3,5,7,9,11]

x1 = [2,4,6,8,10]
y1= [3,6,9,12,15]

plt.bar(x,y)
plt.bar(x1,y1,color = 'g')
```

```python
# Histogram 

a = [1,2,3,4,5,6,7,8,9,2,4,6,8,3,6,9,4,8,5]
plt.hist(a,bins = 5)
```

```python
# Box Plot

d = [np.random.normal(0,std,100) for std in range(1,4)]

plt.boxplot(d,vert = True,patch_artist = True);

#    -->  'o' are outliers
```

```python
# Pie Chart

# Data

label = 'A','B','C','D'
size = [233,443,211,455]
color = ['gold','yellowgreen','lightcoral','blue']
explode = (0,0.06,0,0.09)


plt.pie(size,explode = explode, labels = label, colors = color,shadow = True,autopct = '%1.1f%%')
```
