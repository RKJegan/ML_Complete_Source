positional and keyword arguments in fn

```python
def h(name,age = 20):        #-->  name is positional argument , age = 20  is keyword argument
    print("My name {} and my age {}".format(name,age))
```

```python
h('jegi')
```

```python
h()
```

```python
def h1(name,age):        #-->  name and age are positional argument
    print("My name {} and my age {}".format(name,age))
```

```python
h1('jegi',20)
```

```python
def h2(*args,**kwargs):       #--> args  --> positional argument
    print(args)               #--> kwargs --> keyword argument
    print(kwargs)
```

```python
h2('jegi','jegan',age = 20,dob = 2005)
```

```python
lst_args = ['a','b','e']
dict_kwargs = {'age': 20, 'dob': 2005}
```

```python
h2(*lst_args,**dict_kwargs)      #   use *  and  **  to define args and kwargs
```

return multiple results

```python
lst = [1,2,3,4,5,6,7,8,9]
```

```python
def EOsum(lst):
    es = 0
    os = 0
    for i in lst:
        if i % 2 == 0:
            es += i
        else:
            os += i
    return es,os
```

```python
EOsum(lst)
```
