Map Function

```python
# simple function
```

```python
def eo(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
```

```python
eo(23)
```

```python
l = [1,2,3,4,5,6,7,8,9]
```

```python
map(eo,l)      # --> it creates map address to iterate through function
```

```python
list(map(eo,l))  # --> helps to view map function value
```
