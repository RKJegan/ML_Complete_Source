Lambda Function
(anonymous function)

a function with no name

```python
#  Normal function

def add(a,b): 
    return a+b
```

```python
add(45,77)
```

```python
# Lambda function         used for single expression or operation

add1 = lambda a,b:a+b
```

```python
add1(23,43)
```

```python
even = lambda a:a%2 == 0
```

```python
even(45)
```

```python
add1 = lambda x,y,z : x+y+z
add1(45,66,43)
```
