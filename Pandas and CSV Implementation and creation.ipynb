```python
# CSV File Creation


from io import StringIO, BytesIO

data = ('c1,c2,c3\n'
        'a,b,c\n'
        'd,e,f\n'
        '1,2,3')
```

```python
type(data)
```

```python
pd.read_csv(StringIO(data))
```

```python
# read from specific column

df = pd.read_csv(StringIO(data),usecols = ['c1','c3'])
```

```python
df
```

```python
df.to_csv("str_to_csv.csv")
```

```python
# Specify column data type

data = ('a,b,c,d\n'
        '1,2,3.1,4\n'
        '5,6,7.5,8\n'
        '2,3,4.6,5')
```

```python
print(data)
```

```python
df = pd.read_csv(StringIO(data),dtype = object)     # specify as entire data frame
```

```python
print(type(df))
print(df.info())
df
```

```python
# specify dtype for specific columns

df = pd.read_csv(StringIO(data),dtype = {'a':'int32','b':'Int64','c':float,'d':object})
```

```python
print(df.info())
df
```

```python
# To check data types

df.dtypes
```

```python
pd.read_csv(StringIO(data),index_col = 1)      # set index value as needed
```

```python
pd.read_csv(StringIO(data)) 
```

```python
pd.read_csv(StringIO(data),index_col = False) 
```

```python
# Combine usecols and index_col

pd.read_csv(StringIO(data),usecols = ['a','b'],index_col = False) 
```
