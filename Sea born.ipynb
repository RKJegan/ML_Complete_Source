Seaborn



```python
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
```

```python
df.head(7)
```

Correlation with Heatmap

inter-relation between two features (not on categorical)
value range between -1 to +1

```python
df.corr(numeric_only = True)    #--> simple method to access numerical values for correlation
```

```python
# way to select only numeical values

num_df = df.select_dtypes(include=['number'])
num_df.corr()
```

```python
sns.heatmap(num_df.corr())
```

Join Plot

study relationship between 2 numerical values
central chart display their correlation
it is usually a 
1. scatter plot
2. hexbin plot
3. 2d histogram or 2d density plot

```python
sns.jointplot(x = 'tip', y = 'total_bill',data = df,kind = 'hex')
```

```python
sns.jointplot(x = 'tip', y = 'total_bill',data = df,kind = 'reg')
```

Pair Plot

plot pair-wise relation in the dataset

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue = 'sex')
```

```python
sns.pairplot(df, hue = 'smoker')
```

Dist Plot

see the distribution of the column features

```python
sns.distplot(df['tip'])
```

```python
sns.distplot(df['tip'],kde=False)
```

Categorical Plots

used for categorical data points

1. boxplot
2. violin plot
3. count plot
4. bar plot

```python
# count plot
# Show the counts of observations in each categorical bin using bars.
# use 1 feature like (x or y)


sns.countplot(data = df,x = 'sex')
```

```python
sns.countplot(data = df,x = 'sex',hue = 'smoker')
```

```python
sns.countplot(x = 'day',data = df)
```

```python
sns.countplot(y = 'day',data = df)
```

```python
# Bar Plot
#Show point estimates and errors as rectangular bars.

sns.barplot(x = 'total_bill',y= 'sex',data = df)
```

```python
sns.barplot(y = 'total_bill',x= 'sex',data = df)
```

```python
sns.barplot(y = 'total_bill',x= 'smoker',hue = 'time',data = df)
```

Box Plot

Draw a box plot to show distributions with respect to categories.


```python
# Box Plot

sns.boxplot(x = 'sex',y = 'total_bill',data = df,palette = 'rainbow')
```

```python
sns.boxplot(x = 'day',y = 'total_bill',data = df,palette = 'rainbow')
```

```python
sns.boxplot(data=df,orient = 'v')
```

```python
sns.boxplot(x='day',y = 'total_bill',hue = 'smoker',data=df,palette = 'rainbow')
```

Violin Plot

helps to see both the distribution of ddata in terms of kernel density and the box plot

Draw a patch representing a KDE and add observations or box plot statistics.

```python
# violin plot

sns.violinplot(x = 'day', y = 'total_bill', hue = 'smoker', data = df)
```

IRIS Dataset practice

```python
# load dataset

data = sns.load_dataset("iris")
```

```python
import seaborn as sns
```

Distribution plots

```python
df1 = data.corr(numeric_only = True)
df1
```

```python
# heatmap

sns.heatmap(df1)
```

```python
# distplot

sns.displot(data['sepal_length'])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.subplot(2,2,1)
sns.distplot(data['sepal_length'])
plt.subplot(2,2,2)
sns.distplot(data['sepal_width'])
plt.subplot(2,2,3)
sns.distplot(data['petal_length'])
plt.subplot(2,2,4)
sns.distplot(data['petal_width'])
plt.tight_layout()
plt.show()
```

```python
# displot    Figure-level interface for drawing distribution plots onto a FacetGrid.     updated on distplot

sns.displot(x = 'sepal_length',y = 'sepal_width',data = data,kind='hist')    # kind = kde or hist or ecdf(bi variate)
```

```python
# pair plot

sns.pairplot(data)
```

```python
sns.pairplot(data,hue = 'species')
```

```python
# Joint plot

sns.jointplot(x = 'sepal_length', y = 'petal_length',data= data,kind = 'scatter',hue = 'species')
```

```python
sns.jointplot(x = 'sepal_length', y = 'petal_length',data= data,kind = 'reg')
```

Categorical plots

```python
# Count plot

sns.countplot(x = 'species',data=data)
#plt.tight_layout()
#plt.show()
```

```python
# Box plot

sns.boxplot(y = 'sepal_length', x = 'species',data = data)
```

```python
sns.boxplot(x = 'sepal_length', y = 'sepal_width',data = data,hue='species',palette = 'rainbow')
```

```python
plt.figure(figsize = (6,6))
plt.subplot(2,2,1)
sns.boxplot(y = 'sepal_length', x = 'species',data = data,palette = 'rainbow')
plt.subplot(2,2,2)
sns.boxplot(y = 'sepal_width', x = 'species',data = data,palette = 'rainbow')
plt.subplot(2,2,3)
sns.boxplot(y = 'petal_length', x = 'species',data = data,palette = 'rainbow')
plt.subplot(2,2,4)
sns.boxplot(y = 'petal_width', x = 'species',data = data,palette = 'rainbow')
plt.tight_layout()
```

```python
# violin plot

sns.violinplot(x = 'sepal_length', y = 'species',data = data,palette = 'rainbow')
```

```python
plt.subplot(2,2,1)
sns.violinplot(x = 'sepal_length', y = 'species',data = data,palette = 'rainbow')
plt.subplot(2,2,2)
sns.violinplot(x = 'sepal_width', y = 'species',data = data,palette = 'rainbow')
plt.subplot(2,2,3)
sns.violinplot(x = 'petal_length', y = 'species',data = data,palette = 'rainbow')
plt.subplot(2,2,4)
sns.violinplot(x = 'petal_width', y = 'species',data = data,palette = 'rainbow')
plt.tight_layout()
```

```python
# bar plot

sns.barplot(y = 'sepal_length',x = 'species', data = data)
```
